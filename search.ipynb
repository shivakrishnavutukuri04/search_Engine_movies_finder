Embeddings & FAISS

```python
# Import Required Libraries

import numpy as np
import pandas as pd
import faiss

from sentence_transformers import SentenceTransformer
```

```python
# Load Clean Dataset

df=pd.read_csv(r"movies_dataset.csv")

print("Total Movies :", len(df))
```

```python
df.isnull().sum()
```

```python
df.to_csv(r"movies.csv", index=False)
```

```python
# Load SBERT Model

model = SentenceTransformer("all-MiniLM-L6-v2")

print("Model Loaded Successfully")
```

```python
# Generate Embeddings

movie_embeddings = model.encode(
    df["movie_info"].tolist(),
    batch_size=32,
    show_progress_bar=True,
    convert_to_numpy=True
)
```

```python
print("Embedding Shape :", movie_embeddings.shape)

print("Data Type :", movie_embeddings.dtype)
```

```python
# Save Embeddings

np.save("movie_embeddings.npy", movie_embeddings)

print("Embeddings Saved Successfully")
```

```python
movie_embeddings = np.load("movie_embeddings.npy")

print(movie_embeddings.shape)
```

```python
#Normalize Embeddings

faiss.normalize_L2(movie_embeddings)
```

```python
#Create FAISS Index

dimension = movie_embeddings.shape[1]

index = faiss.IndexFlatIP(dimension)
```

```python
#Add Movie Embeddings

index.add(movie_embeddings)

print("Movies Indexed :", index.ntotal)
```

```python
#Save FAISS Index

faiss.write_index(index, "movie.index")

print("FAISS Index Saved Successfully")
```

```python
#Verify Saved Index
loaded_index = faiss.read_index("movie.index")

print("Indexed Movies :", loaded_index.ntotal)
```

Semantic Search Engine

```python
# Load FAISS Index

index = faiss.read_index("movie.index")
```

```python
# Load SBERT Model

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
# Verify Everything

print("Movies :", len(df))
print("Embeddings :", movie_embeddings.shape)
print("Indexed Movies :", index.ntotal)
```

```python
# Query Preprocessing

def preprocess_query(query):

    query = query.lower()

    query = " ".join(query.split())

    return query
```

```python
# Convert Query to Embedding

def get_query_embedding(query):

    query = preprocess_query(query)

    query_embedding = model.encode(
        [query],
        convert_to_numpy=True
    )

    faiss.normalize_L2(query_embedding)

    return query_embedding
```

```python
# Search Function

def search_movies(query, top_k=3):

    query_embedding = get_query_embedding(query)

    scores, indices = index.search(query_embedding, top_k)

    results = df.iloc[indices[0]].copy()

    results["Similarity Score"] = scores[0].round(3)

    return results.reset_index(drop=True)
```

```python

# Semantic Movie Recommendation Engine
# Hybrid Search


import re
import faiss
import numpy as np
import pandas as pd

from rapidfuzz import fuzz
from rapidfuzz import process

from sentence_transformers import SentenceTransformer
```

```python
print("Movies :", len(df))
print("Embeddings :", movie_embeddings.shape)
print("Indexed Movies :", index.ntotal)
```

```python
def preprocess_query(query):

    query = query.lower()

    query = query.strip()

    query = re.sub(r"\s+", " ", query)

    return query
```

```python
preprocess_query("   BAHUBALI     ")
```

```python
# Fuzzy Title Search# Create a list of movie titles

movie_titles = df["Title"].fillna("").tolist()


```

```python
# Test RapidFuzz

process.extract(
    "bahubali",
    movie_titles,
    scorer=fuzz.WRatio,
    limit=5
)
```

```python
# Build the function

def fuzzy_title_search(query, limit=3):

    query = preprocess_query(query)

    matches = process.extract(
        query,
        movie_titles,
        scorer=fuzz.WRatio,
        limit=limit
    )

    results = []

    for title, score, idx in matches:

        results.append({
            "Index": idx,
            "Title": title,
            "Title Score": score / 100
        })

    return pd.DataFrame(results)
```

```python
fuzzy_title_search("bahubali")
```

```python
fuzzy_title_search("pushpa")
```

```python
def semantic_search(query, limit=10):

    query = preprocess_query(query)

    query_embedding = model.encode(
        [query],
        convert_to_numpy=True
    )

    faiss.normalize_L2(query_embedding)

    scores, indices = index.search(query_embedding, limit)

    results = pd.DataFrame({
        "Index": indices[0],
        "Semantic Score": scores[0]
    })

    return results
```

```python
semantic_search("bahubali")
```

```python
def hybrid_search(query, limit=50):

    title_results = fuzzy_title_search(query, limit)
    semantic_results = semantic_search(query, limit)

    title_results = title_results[["Index", "Title Score"]]
    semantic_results = semantic_results[["Index", "Semantic Score"]]

    results = pd.merge(
        title_results,
        semantic_results,
        on="Index",
        how="outer"
    )

    results["Title Score"] = results["Title Score"].fillna(0)
    results["Semantic Score"] = results["Semantic Score"].fillna(0)

    results["Final Score"] = (
        0.60 * results["Title Score"] +
        0.40 * results["Semantic Score"]
    )

    results = results.sort_values(
        "Final Score",
        ascending=False
    )

    results = results.head(3)

    results["Title"] = df.iloc[results["Index"]]["Title"].values
    results["Year"] = df.iloc[results["Index"]]["Year"].values

    results = results[["Title", "Year"]]

    return results.reset_index(drop=True)
```
