```python
import time
import requests
import pandas as pd
from bs4 import BeautifulSoup


def scrape_telugu_movies(start_year=2000, end_year=2024):
    """
    Scrape Telugu movie details from Wikipedia.

    Returns
    -------
    pandas.DataFrame
    """

    session = requests.Session()

    session.headers.update({
        "User-Agent": (
            "Mozilla/5.0 "
            "(Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 "
            "(KHTML, like Gecko) "
            "Chrome/138.0 Safari/537.36"
        )
    })

    all_movies = []

    for year in range(start_year, end_year + 1):

        print(f"\nScraping {year}...")

        url = f"https://en.wikipedia.org/wiki/List_of_Telugu_films_of_{year}"

        try:

            response = session.get(url, timeout=20)
            response.raise_for_status()

            soup = BeautifulSoup(response.text, "lxml")

            tables = soup.find_all("table", class_="wikitable")

            movies_in_year = 0

            for table in tables:

                # -----------------------------
                # Read Header
                # -----------------------------
                header_row = table.find("tr")

                if header_row is None:
                    continue

                headers = []

                for header in header_row.find_all(["th", "td"]):

                    headers.append(
                        header.get_text(" ", strip=True)
                    )

                # Skip unwanted tables
                if "Title" not in headers and "Film" not in headers:
                    continue

                # -----------------------------
                # Find Column Positions
                # -----------------------------
                title_index = None
                director_index = None
                cast_index = None
                release_index = None

                for i, column in enumerate(headers):

                    column = column.strip()

                    if column in ["Title", "Film"]:
                        title_index = i

                    elif column == "Director":
                        director_index = i

                    elif column == "Cast":
                        cast_index = i

                    elif column in ["Release date", "Release Date"]:
                        release_index = i

                # -----------------------------
                # Read Movie Rows
                # -----------------------------
                rows = table.find_all("tr")[1:]

                for row in rows:

                    cells = row.find_all(["td", "th"])

                    if len(cells) == 0:
                        continue

                    movie = {
                        "Title": None,
                        "Director": None,
                        "Cast": None,
                        "Release Date": None,
                        "Year": year,
                        "Language": "Telugu"
                    }

                    # -----------------------------
                    # Title
                    # -----------------------------
                    if (
                        title_index is not None and
                        title_index < len(cells)
                    ):

                        movie["Title"] = cells[title_index].get_text(
                            separator=" ",
                            strip=True
                        )

                    # -----------------------------
                    # Director
                    # -----------------------------
                    if (
                        director_index is not None and
                        director_index < len(cells)
                    ):

                        movie["Director"] = cells[director_index].get_text(
                            separator=", ",
                            strip=True
                        )

                    # -----------------------------
                    # Cast
                    # -----------------------------
                    if (
                        cast_index is not None and
                        cast_index < len(cells)
                    ):

                        cast_cell = cells[cast_index]

                        actor_links = cast_cell.find_all("a")

                        if actor_links:

                            actors = []

                            for actor in actor_links:

                                actor_name = actor.get_text(strip=True)

                                if actor_name != "":
                                    actors.append(actor_name)

                            movie["Cast"] = ", ".join(actors)

                        else:

                            movie["Cast"] = cast_cell.get_text(
                                separator=", ",
                                strip=True
                            )

                    # -----------------------------
                    # Release Date
                    # -----------------------------
                    if (
                        release_index is not None and
                        release_index < len(cells)
                    ):

                        movie["Release Date"] = cells[
                            release_index
                        ].get_text(
                            separator=" ",
                            strip=True
                        )

                    all_movies.append(movie)
                    movies_in_year += 1

            print(f"Collected {movies_in_year} movies.")

            time.sleep(1)

        except Exception as e:

            print(f"Failed to scrape {year}: {e}")

    # ------------------------------------
    # Create DataFrame
    # ------------------------------------
    movies_df = pd.DataFrame(all_movies)

    # Remove duplicate movies
    movies_df.drop_duplicates(inplace=True)

    # Reset index
    movies_df.reset_index(drop=True, inplace=True)

    print("\nScraping Completed Successfully")
    print("Total Movies :", len(movies_df))

    return movies_df
```
